```python
import math
def get_distance_math(point1, point2):
    # Returns the Euclidean distance between two N-dimensional points
    return math.dist(point1, point2)
```

```python
import collections

#fimpementing knn predictor
de predict(new_HEIGHT,neW_WEIGHT,k=3):
  new_point=[new_HEIGHT,neW_WEIGHT]
  distances=[]
  for i in range(len(x)):
    distance=get_distance_math(new_point,x.iloc[i])
    distances.append((distance,y[i]))
  distances.sort()
  K_nearest=distances[:k]
  K_lables=[lable for(distance,lable) in K_nearest]
  prediction= collections.Counter(K_lables).most_common(1)[0][0]
  return prediction
```
