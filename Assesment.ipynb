```
import matplotlib.pyplot as plt
import numpy as np

# 1. Generate sample original data (Hours vs. Actual Scores)
np.random.seed(42)
hours = np.random.uniform(1, 10, 30)
actual_scores = 5 * hours + 40 + np.random.normal(0, 5, 30)

# 2. Generate sample prediction data (e.g., from a Linear Regression model)
# Using a perfect line equation (y = 5x + 40) as the predicted path
predicted_scores = 5 * hours + 40

# 3. Create the plot container
plt.figure(figsize=(8, 5))

# 4. Plot the original data points
plt.scatter(hours, actual_scores, color='blue', alpha=0.7, label='Original Data')

# 5. Plot the predicted data points on the exact same axes
plt.scatter(hours, predicted_scores, color='red', marker='x', s=60, label='Predictions')

# 6. Customize labels, legend, and grid
plt.title('Actual vs. Predicted Exam Scores by Study Hours')
plt.xlabel('Hours Studied')
plt.ylabel('Exam Score')
plt.legend(loc='upper left')
plt.grid(True, linestyle='--', alpha=0.5)

# 7. Render the final combined chart
plt.show()
```
