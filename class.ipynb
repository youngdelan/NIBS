```
#class bosko
class bosko:
  def __init__(self,name,breed,age):        #this is a function that runs when u create a new bosko
    self.name=name
    self.breed=breed
    self.age=age


  def dogsound(self):
    print(self.name+"woooof woooof")

  def returname(self):#returns to dog name
    print(self.name)
  def returnbreed(self):#returns to dog breed
    print(self.breed)
  def getinfo(self):
    return self.name,self.breed,self.age

bosko1=bosko("steven","pitbul",7)#creating a new dog object
bosko1.dogsound()
bosko1.returname()
bosko1.returnbreed()

bosko=bosko1.getinfo()#storing the values of getinfo method to a variable
print(bosko)
```
