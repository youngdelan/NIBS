```
#using an init function
class car:
  def __init__(self,name,year):
    self.name=name
    self.year=year
  carObj=car("BMW-M4",2021)
  print(carObj.name)
  print(carObj.year)
```

```
#inheritance overridding
class animal:
  def __init__(self,name,age):
    self.name=name
    self.age=age
  def eat(self):
      print(f"{self.name} i love eating bones")
  def sound(self):
    print(f"{self.name} barks loudly")
class dog(animal):
  def bark(self):
    print(f"{self.name} fluffy barks loudly")
animal1=dog("steve",5)
animal1=dog("steve",5)
animal1.bark()
print(animal1.eat())
print(animal1.bark())
print(animal1.sound())

```
