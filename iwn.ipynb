```
#program that represents student score and displays competency
score=int(input("enter student score;"))
if score <0:
 print("error")
elif score>=90:
 print("excelent")
elif score>=80:
 print("good")
elif score>=70:
 print("average")
elif score>=60:
 print("pass")
else:
 print("fail")
```

```
#polymophism
class cars():
  def type(self):
    print("types of cars")
class motobike():
  def type(toyota):
    print("kawasaki")
class vehicle():
  def type(ford):
    print("F450")
class lorry():
  def type(isuzu):
    print("MAN")
cars=[cars(),motobike(),vehicle(),lorry()]
print(cars[0].type())
print(cars[1].type())
print(cars[2].type())
print(cars[3].type())
```
