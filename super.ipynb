```
#classes
class vehicle:
  def __init__(self,brand,year_of_manufacture,color):
    self.brand=brand      #parent class/base class/super class
    self.year_of_manufacture=year_of_manufacture
    self.color=color
  def move(self):
    super().move()
    print("vehicle is moving")          #introduning super method to ensure parent class is not over ridden
class car(vehicle):
  def move(self):
    print("driving on the road")
class bike(vehicle):
  def move(self):
    print("pedalling on the path")


carObj=car("2025","subaru","blue")
carObj.move()

bikeObj=bike("2019","kawasaki","green")
bikeObj.move()
```
